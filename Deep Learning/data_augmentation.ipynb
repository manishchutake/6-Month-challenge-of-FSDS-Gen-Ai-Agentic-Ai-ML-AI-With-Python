```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img, img_to_array
```

```python
#create an image data generator with augmentations
datagen = ImageDataGenerator(
    rotation_range = 40,
    width_shift_range = 0.4,
    height_shift_range = 0.4,
    shear_range = 0.4,
    zoom_range = 0.4,
    horizontal_flip = True,
    rescale = None,
    fill_mode = 'wrap' #nearest, reflect , wrap constant we can use this option also
    
)
```

```python
#load image from path
img  = load_img(r"/Users/manishuxuid/Desktop/Finale Practise/Deep Learning/Data Augmentarion/indian-man-smiling-mockup-psd-cheerful-expression-closeup-portra.jpg")
```

```python
img
```

```python
#convert numpy to array
x = img_to_array(img)
x = x.reshape((1,) + x.shape) #reshape to(1,height, width, channels)
#generate augmented images and save
i = 0
for batch in datagen.flow(x, batch_size =1,
                          save_to_dir = (r"/Users/manishuxuid/Desktop/Finale Practise/Deep Learning/Data Augmentarion/Data Augment"),
                          save_prefix = 'man',save_format = 'jpeg'):
    i +=1
    if i > 30:
        break #prevent infinite loop
    
```
